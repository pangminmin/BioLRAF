# BioLRAF visualization example: GSE226365

This notebook demonstrates the BioLRAF 3D ternary visualization workflow using the complete GSE226365 dataset. The same cells are plotted twice: first colored by `celltype`, and then colored by `lineage_dominant`. The input must be a cell-level score table containing `Os`, `Ch`, `Ad`, `MSC`, and `celltype`. It is not the gene-level gficf matrix exported immediately after preprocessing.

```python
from pathlib import Path
import sys
import pandas as pd

CURRENT_DIR = Path.cwd().resolve()
METHODS_DIR = CURRENT_DIR if CURRENT_DIR.name == "Methods" else CURRENT_DIR / "Methods"
if not (METHODS_DIR / "BioLRAF_visualization.py").exists():
    raise FileNotFoundError(
        "Cannot find Methods/BioLRAF_visualization.py. Start Jupyter from the BioLRAF repository root or its Methods directory."
    )
REPO_ROOT = METHODS_DIR.parent
sys.path.insert(0, str(METHODS_DIR))

from BioLRAF_visualization import (
    calculate_biolraf_metrics,
    export_total_dataset_plots,
    read_score_table,
    safe_filename,
)
```

## Configuration

Adjust the relative paths if the repository structure differs from the example layout.

```python
INPUT_FILE = REPO_ROOT / "Example" / "GSE226365" / "input" / "GSE226365_BioLRAF_scores.csv"
OUTPUT_DIR = REPO_ROOT / "Example" / "GSE226365" / "results"
DATASET_LABEL = "GSE226365"
CENTER_DELTA = 0.10
```

## Calculate BioLRAF metrics and dominant lineage states

```python
scores = read_score_table(INPUT_FILE)
scores = calculate_biolraf_metrics(scores, center_delta=CENTER_DELTA)

display(scores.head())
display(scores["celltype"].value_counts())
display(pd.crosstab(scores["celltype"], scores["lineage_dominant"]))
```

## Export total-dataset plots

This step generates one plot colored by `celltype` and one plot colored by `lineage_dominant`. Each plot is saved as interactive HTML and four fixed PNG views.

```python
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
processed_file = OUTPUT_DIR / f"{safe_filename(DATASET_LABEL)}_BioLRAF_scores_with_states.csv"
scores.to_csv(processed_file, index=False)
print(f"Saved: {processed_file}")

export_total_dataset_plots(
    df=scores,
    output_dir=OUTPUT_DIR,
    dataset_label=DATASET_LABEL,
)
```
